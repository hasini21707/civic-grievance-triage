# 04. K-Means Ward Recurrence Analysis\nEvaluating ward complaint density and clustering 199 BBMP wards into Low, Medium, and High Recurrence Risk.

```
import sys
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sys.path.append('..')

from src.recurrence_model import WardRecurrenceAnalyzer

df = pd.read_csv('../data/processed/cleaned_grievances.csv', low_memory=False)

analyzer = WardRecurrenceAnalyzer(n_clusters=3, random_state=42)
ward_rec_df = analyzer.fit_and_cluster(df)

print('=== WARD RECURRENCE RISK SUMMARY ===')
print(ward_rec_df['recurrence_level'].value_counts())
print('\nTop 10 Recurrence Risk Wards:')
print(ward_rec_df[['Ward_Name_Clean', 'complaint_count', 'recurrence_score', 'recurrence_level']].head(10))
```

### Elbow Method & Silhouette Score Plots

```
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(list(analyzer.elbow_scores.keys()), list(analyzer.elbow_scores.values()), 'bo-')
plt.title('Elbow Method (Inertia vs k)')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')

plt.subplot(1, 2, 2)
plt.plot(list(analyzer.silhouette_scores.keys()), list(analyzer.silhouette_scores.values()), 'ro-')
plt.title('Silhouette Score vs k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Silhouette Score')
plt.tight_layout()
plt.show()
```
